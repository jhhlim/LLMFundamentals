## Setup — Colab AI models

This homework follows class lab 4a and uses `google.colab.ai` (run in Google Colab).


```python
# @title List available models
from google.colab import ai

ai.list_models()
```

```python
# Default model for this homework (Flash = fast for prompt experiments)
MODEL = "google/gemini-2.5-flash-lite"

def ask(prompt, model_name=MODEL, stream=True):
    """Send a prompt via Colab AI and print the response."""
    from google.colab import ai
    if stream:
        for text in ai.generate_text(prompt, model_name=model_name, stream=True):
            print(text, end="")
        print()
    else:
        print(ai.generate_text(prompt, model_name=model_name))
```

---
## Style #1 — Zero-Shot Prompting

**Core idea:** Ask the model to do a task directly, with **no examples**.  
**When to use:** Simple Q&A, quick brainstorming, when the model already knows the domain.

**Theme prompt:** Explain how LLMs are used in robotics (direct ask, with a light audience cue).


```python
zero_shot_prompt = """
Explain how large language models (LLMs) are used in robotics today.
Cover three areas: (1) natural-language robot control, (2) task planning,
and (3) human-robot collaboration. Keep the answer under 150 words and
write it for a university engineering student.
""".strip()

print("=== ZERO-SHOT PROMPT ===")
print(zero_shot_prompt)
print("\n=== MODEL RESPONSE ===")
ask(zero_shot_prompt)
```

---
## Style #2 — Few-Shot Prompting

**Core idea:** Give a few **input → output examples** so the model learns the desired format.  
**When to use:** Classification, extraction, or a specific short output style.

**Theme prompt:** Classify robotics LLM use-cases into short category labels, taught with examples.


```python
few_shot_prompt = """
System: You classify how an LLM is being used in a robotics scenario.
Use exactly one label from: Language-to-Action, Task-Planning, Perception-Captioning, Code-Generation, HRI-Dialogue.
Reply with only the label.

User: A warehouse robot receives the spoken command "pick the red box on shelf B" and converts it into grasp + navigate actions.
Assistant: Language-to-Action

User: An LLM breaks a goal "assemble the phone charger" into ordered subtasks for a robot arm.
Assistant: Task-Planning

User: A robot camera feed is described in natural language so operators understand what the robot sees.
Assistant: Perception-Captioning

User: A technician asks a cobot why it paused near the conveyor, and the LLM explains the safety stop in plain English.
""".strip()

print("=== FEW-SHOT PROMPT ===")
print(few_shot_prompt)
print("\n=== MODEL RESPONSE ===")
ask(few_shot_prompt)
```

---
## Style #3 — Role Prompting

**Core idea:** Tell the LLM to adopt a **persona / expert role**.  
**When to use:** Expert explanations, audience-specific voice, simulated specialist advice.

**Theme prompt:** Answer as a robotics research lead explaining LLM use on the factory floor.


```python
role_prompt = """
System: You are Dr. Maya Okonkwo, Director of Robot Learning at a major automation lab.
You have 20 years of experience deploying industrial robots and researching LLM-driven
robot policies. You speak clearly, use real-world factory examples, and always mention
both benefits and safety constraints.

User: Dr. Okonkwo, how are LLMs actually used with robots on a modern factory floor?
Give a practical overview an operations manager could understand.
""".strip()

print("=== ROLE PROMPT ===")
print(role_prompt)
print("\n=== MODEL RESPONSE ===")
ask(role_prompt)
```

---
## Style #5 — Emotion Prompting

**Core idea:** Ask the model to write with a **specific emotion / emotional perspective**.  
**When to use:** Empathetic messaging, persuasive tone, emotionally colored creative writing.

**Theme prompt:** A hopeful, inspiring note about how LLMs help people work safely with robots.


```python
emotion_prompt = """
Write a short message (5–7 sentences) about how LLMs are used in robotics to help
human workers collaborate with robots more safely and confidently.

Make it sound warmly optimistic and encouraging — as if speaking to a nervous new
factory technician on their first day with a collaborative robot. Convey hope and
reassurance, not hype. End with one concrete everyday example of LLM + robot teamwork.
""".strip()

print("=== EMOTION PROMPT ===")
print(emotion_prompt)
print("\n=== MODEL RESPONSE ===")
ask(emotion_prompt)
```

---
## Style #7 — Chain-of-Thought (CoT) Prompting

**Core idea:** Ask the model to **think step by step** before the final answer.  
**When to use:** Multi-step reasoning, planning, debugging complex answers.

**Theme prompt:** Reason through which LLM robotics capabilities fit a hospital delivery robot scenario (zero-shot CoT).


```python
cot_prompt = """
Scenario: A hospital wants a mobile robot that can:
A) understand nurse requests like "bring saline to Room 412",
B) plan a route that avoids surgery corridors during procedures, and
C) explain to staff why a delivery was delayed.

Question: Which three LLM-in-robotics capabilities are most important here,
and in what order should the robot software team implement them? Justify the order.

Let's think step by step.
""".strip()

print("=== CHAIN-OF-THOUGHT PROMPT ===")
print(cot_prompt)
print("\n=== MODEL RESPONSE ===")
ask(cot_prompt)
```

| Style | What it controlled in the robotics theme |

| **#1 Zero-Shot** | Direct overview — no demos needed; good for a baseline answer |
